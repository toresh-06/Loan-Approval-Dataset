```python
# cleaning the dataset , converting totalcharges to numeric values

df['totalcharges']=pd.to_numeric(df['totalcharges'],errors='coerce')
```

```python
#Droping the missing rows

df=df.dropna()
```

```python
df.shape
```

```python
# Encoding the target Variable
df['churn']=df['churn'].map({'Yes':1,'No':0})
```

```python
# feature(x) and target(y) separation
x=df.drop(['churn','customerid'],axis=1)
y=df['churn']
```

```python
cat_cols=x.select_dtypes(include='object').columns
num_cols=x.select_dtypes(exclude='object').columns

cat_cols,num_cols
```

```python
# Train and Test Split

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer

x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42,stratify=y)

print("Shape of x_train:", x_train.shape)
print("Shape of x_test:", x_test.shape)
print("Shape of y_train:", y_train.shape)
print("Shape of y_test:", y_test.shape)
```

```python
y.value_counts(normalize=True)
```

```python
y_train.value_counts(normalize=True)
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        ('cat', OneHotEncoder(handle_unknown='ignore'), cat_cols),
        ('num', 'passthrough', num_cols)
    ])

# x_train_encoded = preprocessor.fit_transform(x_train)
# x_test_encoded = preprocessor.transform(x_test)

# print("Shape of x_train_encoded:", x_train_encoded.shape)
# print("Shape of x_test_encoded:", x_test_encoded.shape)
```

```python
# Now we will create a pipeline end to end for OneHotEncoder

# one object, no leakage

from sklearn.pipeline import Pipeline
from sklearn.tree import DecisionTreeClassifier
```

```python
# pipeline = prepocessing + model together
# this avoid the mess up in order

df_pipeline=Pipeline(steps=[
    ('preprocessor',preprocessor),
    ('model',DecisionTreeClassifier(random_state=42))
])
```

```python
# Train the Model with decision tree

df_pipeline.fit(x_train,y_train)
```

```python
#Evaluating the model

from sklearn.metrics import accuracy_score,confusion_matrix,classification_report
y_pred_test=df_pipeline.predict(x_test)
print('Test accuracy_score',accuracy_score(y_test,y_pred_test))

y_pred_train=df_pipeline.predict(x_train)
print('Train accuracy_score',accuracy_score(y_train,y_pred_train))

# if train acc is high , test acc is low - overfitting
# if train acc is low , test acc is high - underfitting
# if train and test acc is high, good model and same -properfitting
```

```python
print(confusion_matrix(y_test,y_pred_test))
```

```python
#classification_report
print(classification_report(y_test,y_pred_test))
```

```python
for depth in [2,4,6,10,None]:
  df_pipeline=Pipeline(steps=[
    ('preprocessor',preprocessor),
    ('model',DecisionTreeClassifier(max_depth=depth,random_state=42))
  ])
  df_pipeline.fit(x_train,y_train)
  y_pred_train = df_pipeline.predict(x_train)
  train_acc = accuracy_score(y_train, y_pred_train)

  y_pred_test = df_pipeline.predict(x_test)
  test_acc = accuracy_score(y_test, y_pred_test)

  print(f'max_depth={depth}, Train={train_acc:.2f}, Test={test_acc:.2f}')
```

```python
df_model=df_pipeline.named_steps['model']
encoder=df_pipeline.named_steps['preprocessor']

feature_names=encoder.get_feature_names_out()

importances=pd.Series(df_model.feature_importances_,index=feature_names)

importances.sort_values(ascending=False,inplace=True)

importances.head(10)
```

```python
from sklearn.ensemble import RandomForestClassifier

rf=Pipeline(steps=[
    ('preprocessor',preprocessor),
    ('model',RandomForestClassifier(n_estimators=200,random_state=42,max_depth=8,n_jobs=-1))
])

rf.fit(x_train,y_train)
```

```python
y_pred_train_rf = rf.predict(x_train)
print('Train accuracy_score for Random Forest:', accuracy_score(y_train, y_pred_train_rf))

y_pred_test_rf = rf.predict(x_test)
print('Test accuracy_score for Random Forest:', accuracy_score(y_test, y_pred_test_rf))
```

```python
#Gradient Boosting

from sklearn.ensemble import GradientBoostingClassifier

gb_pipeline=Pipeline([
    ('preprocessor',preprocessor),
    ('model',GradientBoostingClassifier(n_estimators=150,random_state=42,max_depth=3,learning_rate=0.1))
])

gb_pipeline.fit(x_train,y_train)
```

```python
y_pred_train_gb = gb_pipeline.predict(x_train)
print('Train accuracy_score for Gradient Boosting:', accuracy_score(y_train, y_pred_train_gb))

y_pred_test_gb = gb_pipeline.predict(x_test)
print('Test accuracy_score for Gradient Boosting:', accuracy_score(y_test, y_pred_test_gb))
```
